```python
import re
from datasets import load_dataset, Dataset
import ast
import concurrent
import os
import multiprocessing

# Load and prep dataset
SYSTEM_PROMPT = """
Generate a prolog solution for the asked question.
Follow these steps to craft your response:
1. reason about the given instruction
2. provide a high-quality prolog solution
3. write a query to verify the solution.
Output in the following format:
<reasoning>
...
</reasoning>
<code>
...
</code>
<query>
...
</query>

Write the query just inside <query></query> not in <code></code>.
"""
timeout = 3

def extract_xml_knowledge(text: str) -> str:
    answer = text.split("<code>")[-1]
    answer = answer.split("</code>")[0]
    return answer.strip()

def extract_xml_query(text: str) -> str:
    answer = text.split("<query>")[-1]
    answer = answer.split("</query>")[0]
    return answer.strip()

def split_prolog_rules(code):
    """
    Splits a string of multiple Prolog facts and rules by matching
    periods that are not part of a floating-point number. It removes
    any surrounding whitespace from each rule.
    """
    # This regex splits on a period that is NOT immediately followed by a digit.
    # \s* optionally matches any whitespace before and after the period.
    rules = re.split(r'\s*\.(?!\d)\s*', code)
    # Filter out any empty strings that may occur.
    return [rule for rule in rules if rule]

def parse_kb(prolog_code, query, answer):
    reward = 0
    from pyswip import Prolog
    try:
        prolog_interpreter = Prolog()
        # Filter out any empty rule
        #rules = [rule.strip() for rule in prolog_code.split(").") if rule.strip()]
        #rules = [rule + ")" for rule in rules]
        rules = split_prolog_rules(prolog_code)
        print("-"*20)
        print("rules: ", rules)
        print("-"*20)
        for rule in rules:
            print("Asserting rule: ", rule)
            with concurrent.futures.ThreadPoolExecutor() as executor:
                print("before future")
                future = executor.submit(prolog_interpreter.assertz(rule))
                print("after future")
                try:
                    print("Before result")
                    result = future.result(timeout=timeout)
                    #result = list(future.result(timeout=timeout))
                except concurrent.futures.TimeoutError:
                    print("Query timed out")
                    return 0, None
        
        # Use the instance's query method
        print("Querying: ", query)
        with concurrent.futures.ThreadPoolExecutor() as executor:
            future = executor.submit(prolog_interpreter.query, query)
            try:
                result = future.result(timeout=timeout)
                #result = list(future.result(timeout=timeout))
            except concurrent.futures.TimeoutError:
                print("Query timed out")
                return 0, None
        print("Result: ", result)
        
        for inference in result:
            print("Inference: ", inference)
            for _, result_inference in inference.items():
                print("result_inference: ", result_inference)
                print("Infered: {}, Response: {}, Match: {}".format(result_inference, answer, float(result_inference) == float(answer)))
                try:
                    if float(result_inference) == float(answer):
                        print("Matching!")
                        return 1, result_inference
                except:
                    print("Matching error!")
                    return 0, None
        #print(result)
        # Ensure that the comparison makes sense:
        # This assumes you expect a non-empty result when the answer is correct.
        return 0, None
    except Exception as e:
        print(f"Error encountered: {e}")
        return 0, None

def remove_prolog_comments_and_whitespace(code):
    # Remove block comments (/* ... */)
    code_no_block = re.sub(r'/\*[\s\S]*?\*/', '', code)
    # Remove single-line comments (% ...) from each line
    code_no_comments = re.sub(r'(?m)%.*$', '', code_no_block)
    # Remove newline and tab characters, but keep spaces
    cleaned_code = re.sub(r'[\n\t]', '', code_no_comments)
    return cleaned_code
```

```python
# Query the model using data/prolog_tasks_ground_truth.csv
# The dataset contains the following columns:
# Unnamed:, task_name, task_description, code, test_input_1, test_input_2, ground_truth_1, ground_truth_2
import pandas as pd
dataset = pd.read_csv("data/prolog_tasks_ground_truth.csv")
display(dataset.head())

from vllm import SamplingParams
sampling_params = SamplingParams(
    temperature = 0,
    top_p = 0.000001,
    max_tokens = 1024,
)

results_df = pd.DataFrame(columns=['task_name', 'code', 'match_1', 'match_2', 'test_input_1', 'test_input_2', 'result_1', 'result_2', 'ground_truth_1', 'ground_truth_2'])

def process_row(row):
    total_matches = 0
    user_prompt = row[1]['task_description']
    user_prompt = user_prompt + "Query the function with ?- " + row[1]['test_input_1']
    print("-"*20)
    print("USER PROMPT: ", user_prompt)
    print("-"*20)
    prompt = [{'role': 'system', 'content': SYSTEM_PROMPT},{'role': 'user', 'content': user_prompt}]

    text = tokenizer.apply_chat_template(prompt, tokenize = False, add_generation_prompt = True)

    output = model.fast_generate(
        [text],
        sampling_params = sampling_params,
        lora_request = None,
    )[0].outputs[0].text

    tmp_code = extract_xml_knowledge(output)
    tmp_code = remove_prolog_comments_and_whitespace(tmp_code)


    #tmp_code = row[1]['code']


    print("-"*20)
    print("CODE: ", tmp_code)
    match_1 = 0
    match_2 = 0
    result_1 = None
    result_2 = None
    for i in range(1, 3):
        # tmp_query = extract_xml_query(output)
        # print("-"*20)
        # print("QUERY: ", tmp_query)
        tmp_query = row[1]['test_input_' + str(i)]
        tmp_answer = row[1]['ground_truth_' + str(i)]
        # tmp_answer is a string in the following format "[{'F': 6765}]". Extract the list as [{'F': 6765}]
        tmp_answer = ast.literal_eval(tmp_answer)
        for inference in tmp_answer:
            for _, result_inference in inference.items():
                tmp_answer = result_inference
        print("-"*20)
        print("ANSWER: ", tmp_answer)
        print("-"*20)
        try:
            if i == 1:
                match_1, result_1 = parse_kb(tmp_code, tmp_query, tmp_answer)
            else:
                match_2, result_2 = parse_kb(tmp_code, tmp_query, tmp_answer)
        except Exception as e:
            print("Error encountered: ", e)
    results_df.loc[len(results_df)] = [row[1]['task_name'], tmp_code, match_1, match_2, row[1]['test_input_1'], row[1]['test_input_2'], result_1, result_2, row[1]['ground_truth_1'], row[1]['ground_truth_2']]
    return total_matches

matches = 0
for row in dataset.iterrows():
    print("Processing row: ", row)
    print("Row[1]['task_name']: ", row[1]['task_name'])
    matches += process_row(row)
print("Total matches: ", matches)
```

```python
import pandas as pd
from pyswip import Prolog
import tempfile
import os

def run_prolog_test(code, test_input):
    with tempfile.NamedTemporaryFile(suffix='.pl', mode='w', delete=False) as temp_file:
        temp_file_path = temp_file.name
        temp_file.write(code)
    
    try:
        prolog = Prolog()
        prolog.consult(temp_file_path)
        
        query_results = list(prolog.query(test_input))
        if not query_results:
            return "false"
        elif query_results == [{}]:
            return "true"
        else:
            return str(query_results)
            
    except Exception as e:
        return f"Error: {str(e)}"
    finally:
        os.unlink(temp_file_path)

def test_all_prolog_code(data_path):
    df = pd.read_csv(data_path)
    
    df['result_1'] = None
    df['result_2'] = None
    df['test_1_passed'] = None
    df['test_2_passed'] = None
    
    for idx, row in df.iterrows():
        print(f"Testing task {idx+1}/{len(df)}: {row['task_name']}")
        
        result_1 = run_prolog_test(row['code'], row['test_input_1'])
        result_2 = run_prolog_test(row['code'], row['test_input_2'])
        
        df.at[idx, 'result_1'] = result_1
        df.at[idx, 'result_2'] = result_2
        

        def normalize_result(value):
            value_str = str(value).lower()
            
            # Handle error cases
            if value_str.startswith("error"):
                return "error"
                
            # Convert Boolean results
            if "result" in value_str and "false" in value_str:
                return "false"
            if "result" in value_str and "true" in value_str:
                return "true"
            if value_str == "true" or value_str == "\"true\"":
                return "true"
            if value_str == "false" or value_str == "\"false\"":
                return "false"
            if value_str == "{true}" or value_str == "[{true}]":
                return "true"
                
            # Normalize dictionary keys in list results
            # Convert patterns like [{'X': 24}] to [{'x': 24}]
            if "[{'" in value_str and "'}]" in value_str:
                # Extract the key
                key_start = value_str.find("[{'") + 3
                key_end = value_str.find("':", key_start)
                if key_start != -1 and key_end != -1:
                    key = value_str[key_start:key_end].lower()
                    # Replace the key with lowercase
                    value_str = value_str.replace(value_str[key_start:key_end], key)
                    
            # Handle LCS case specifically
            if "lcs" in value_str:
                # Normalize spacing in LCS lists
                value_str = value_str.replace(" ", "")
                
            return value_str
            
        norm_ground_truth_1 = normalize_result(row['ground_truth_1'])
        norm_ground_truth_2 = normalize_result(row['ground_truth_2'])
        norm_result_1 = normalize_result(result_1)
        norm_result_2 = normalize_result(result_2)
        
        print(f"Task: {row['task_name']}")
        print(f"GT1: {row['ground_truth_1']} -> {norm_ground_truth_1}")
        print(f"Result1: {result_1} -> {norm_result_1}")
        print(f"Match: {norm_result_1 == norm_ground_truth_1}")
        print("---")
        df.at[idx, 'test_1_passed'] = norm_result_1 == norm_ground_truth_1
        df.at[idx, 'test_2_passed'] = norm_result_2 == norm_ground_truth_2

    return df
```
