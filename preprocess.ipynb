```python
# Function to extract descriptions and code from original dataframe
def create_prolog_tasks_dataframe(df_prolog, selected_tasks, test_inputs):
    # Initialize lists to store data
    data = []
    
    # Loop through selected tasks
    for task in selected_tasks:
        # Try to find the task in the dataframe
        task_row = df_prolog[df_prolog['task_name'] == task]
        
        if not task_row.empty:
            # Get description and code from the dataframe
            description = task_row['task_description'].values[0]
            code = task_row['code'].values[0] if 'code' in df_prolog.columns else "Code not available"
            
            # Get test inputs
            test_input_1 = test_inputs[task][0] if task in test_inputs else ""
            test_input_2 = test_inputs[task][1] if task in test_inputs else ""
            
            # Add to data list
            data.append({
                'task_name': task,
                'task_description': description,
                'code': code,
                'test_input_1': test_input_1,
                'test_input_2': test_input_2
            })
        else:
            # Task not found in dataframe
            print(f"Warning: Task '{task}' not found in the dataframe")
            
            # Add with empty description and code
            data.append({
                'task_name': task,
                'task_description': "Description not available",
                'code': "Code not available",
                'test_input_1': test_inputs[task][0] if task in test_inputs else "",
                'test_input_2': test_inputs[task][1] if task in test_inputs else ""
            })
    
    # Create DataFrame
    return pd.DataFrame(data)
```

```python
dataset = create_prolog_tasks_dataframe(df_prolog, selected_tasks, test_inputs)
dataset.to_csv('dataset.csv')
```

```python
from pyswip import Prolog
import pandas as pd
import re
```

```python
def split_prolog_rules(code):
    """
    Splits a string of multiple Prolog facts and rules by matching
    periods that are not part of a floating-point number. It removes
    any surrounding whitespace from each rule.
    """
    rules = re.split(r'\s*\.(?!\d)\s*', code)
    return [rule for rule in rules if rule]
```

```python
inter = Prolog()
fibonacci_code = """
fib(1, 1):- !.
fib(0, 0):- !.
fib(N, Value):-
  A is N - 1, fib(A, A1),
  B is N - 2, fib(B, B1),
  Value is A1 + B1.
"""

rules = split_prolog_rules(fibonacci_code)
for rule in rules: 
  inter.assertz(rule)

result = list(inter.query("fib(10,F)"))
result
```

```python
from pyswip import Prolog
import pandas as pd
import re

df = pd.read_csv('data/dataset.csv')

# Add columns for results
df['ground_truth_1'] = ""
df['ground_truth_2'] = ""

for idx, row in df.iterrows():
    task_name = row['task_name']
    code = row['code']
    test_input_1 = row['test_input_1']
    test_input_2 = row['test_input_2']
    
    print(f"Testing {task_name}...")
    
    # Skip if no code
    if code == "Code not available" or pd.isna(code):
        df.at[idx, 'ground_truth_1'] = "No code available"
        df.at[idx, 'ground_truth_2'] = "No code available"
        continue
    
    # Initialize Prolog once per task
    inter = Prolog()
    
    # Split code into rules
    rules = re.split(r'\s*\.\s*(?!\d)', code)
    
    # Load all rules at once
    for rule in rules:
        if rule.strip():
            inter.assertz(rule)
```
