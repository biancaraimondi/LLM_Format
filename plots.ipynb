```python
# read csv
import csv
import os
import pandas as pd

# from "results" repository read all csv files
results_dir = "results"

all_csv = []
# for all csv file in results_dir or results_dir/subdir or results_dir/subdir/subdir
for root, dirs, files in os.walk(results_dir):
    for file in files:
        if file.endswith(".csv"):
            csv_file = os.path.join(root, file)
            all_csv.append(csv_file)
all_csv = [csv_file for csv_file in all_csv if "gsm_symbolic" in csv_file and "results.csv" not in csv_file and "one_shot" in csv_file and
"p1" in csv_file]
print(all_csv)

# create dictionary with all csv files
all_data = {}
for csv_file in all_csv:
    if os.path.isfile(f"{csv_file}"):
        data = pd.read_csv(f"{csv_file}")
        all_data[csv_file] = data
```

```python
# from all_data dictionary composed by name: data, create a table with the data calculating accuracy by counting rows with mean > 0
df = pd.DataFrame({"name": all_data.keys(),
                   "accuracy": [all_data[key]["match"].sum()/len(all_data[key]) for key in all_data.keys()],
                   "strict_accuracy": [(all_data[key]["sum"] == 4).sum()/len(all_data[key]) for key in all_data.keys()]
                   })
display(df)

# save the table to csv
df.to_csv("results/results.csv", index=False)
```

#### Plots for Rosetta Code

```python
import os
import pandas as pd

results_dir = "results"
all_csv = []
for root, dirs, files in os.walk(results_dir):
    for file in files:
        if file.endswith(".csv"):
            csv_file = os.path.join(root, file)
            all_csv.append(csv_file)
print(all_csv)

rows_to_remove = [6, 12, 13, 14, 18]
all_data = {}
for csv_file in all_csv:
    if "rosetta2" in os.path.basename(csv_file).lower() and os.path.isfile(csv_file):
        data = pd.read_csv(csv_file)
        if len(data) > 0:
            filtered_data = data.drop(index=[i for i in rows_to_remove if i < len(data)])
            all_data[csv_file] = filtered_data
```
