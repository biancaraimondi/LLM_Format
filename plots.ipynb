```python
# read csv
import csv
import os
import pandas as pd

# from "results" repository read all csv files
results_dir = "results"
results_files = os.listdir(results_dir)
all_csv = []
for i in range(len(results_files)):
    for j, csv_file in enumerate(os.listdir(results_files[i])):
        all_csv.append(f"{results_dir}/{results_files[i]}/{csv_file}.csv")
print(all_csv)
# create dictionary with all csv files
all_data = {}
for csv_file in all_csv:
    if os.path.isfile(f"{csv_file}"):
        data = pd.read_csv(f"{csv_file}")
        all_data[csv_file] = data
```

```python
# from all_data dictionary composed by name: data, create a table with the data calculating accuracy by counting rows with mean > 0
df = pd.DataFrame({"name": all_data.keys(), "accuracy": [all_data[key]["match"].sum()/len(all_data[key]) for key in all_data.keys()]})
display(df)
```
